```python
# Customer Churn & Retention Analytics
# Python: Data Cleaning & Basic Churn Analysis

import pandas as pd
import numpy as np

# 1. Load dataset
df = pd.read_csv("Telco-Customer-Churn.csv")

# 2. Understand the dataset
print("Shape:", df.shape)

print("\nColumns:")
print(df.columns.tolist())

print("\nMissing values:")
print(df.isna().sum())

print("\nDuplicate rows:", df.duplicated().sum())

# 3. Clean TotalCharges
df["TotalCharges"] = pd.to_numeric(
    df["TotalCharges"],
    errors="coerce"
)

df["TotalCharges"] = df["TotalCharges"].fillna(0)

# 4. Validate cleaned data
print("\nTotalCharges data type:")
print(df["TotalCharges"].dtype)

print("\nMissing values after cleaning:")
print(df.isna().sum())

# 5. Basic churn analysis
print("\nChurn distribution:")
print(df["Churn"].value_counts())

churn_rate = (
    df["Churn"].value_counts()["Yes"] / len(df)
) * 100

print("\nChurn Rate:", round(churn_rate, 2), "%")

# 6. Save cleaned dataset
df.to_csv("customer_churn_clean.csv", index=False)

print("\nCleaned dataset saved successfully.")
```
